$$\int \frac{dx}{x\sqrt{5-x^2}}$$
$$x = \sqrt{5}\cos{\theta} \quad x \in [-\sqrt{5}, \sqrt{5}]$$
$$\frac{1}{\sqrt{5}}\int \frac{-\sin{\theta}d\theta}{\cos{\theta}\sqrt{1-\cos^2{\theta}}}$$
$$-\frac{1}{\sqrt{5}}\int \frac{d\theta}{\cos{\theta}} = -\frac{1}{\sqrt{5}}\int \frac{\cos{\theta}d\theta}{\cos^2{\theta}} = -\frac{1}{\sqrt{5}}\int \frac{d(\sin\theta)}{1 - \sin^2\theta} = -\frac{1}{2\sqrt{5}}\ln{\left|\frac{1 + \sin\theta}{1-\sin\theta}\right|}$$

$$\sin\theta = \sqrt{1 - \frac{x^2}{5}}$$

Ответ:
$$\frac{1}{2\sqrt{5}}\ln{\left|\frac{\sqrt{5}-\sqrt{5 - x^2}}{\sqrt{5} + \sqrt{5 - x^2}}\right|}$$
