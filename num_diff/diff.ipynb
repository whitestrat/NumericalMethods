Производная через интерполяцию

$$f(x) \approx L(x) = \sum_{k = i-n}^{k = i + n} f(x_k) \prod_{\substack{j = i-n \\ j \neq k}}^{j = i + n}{\frac{x - x_j}{x_k - x_j}}$$
$$L'(x) = \sum_{k = i-n}^{k = i + n} f(x_k) \sum_{\substack{j = i-n \\ j \neq k}}^{j = i + n} \frac{1}{x_k - x_j} \prod_{\substack{j = i-n \\ l \neq k \\ l \neq j}}^{l = i+n}{\frac{x - x_l}{x_k - x_l}}$$

 Вторая производная через центральные разности

 $O(h^2) :$
 $$
 \begin{cases}
    f(x + h) = f(x) + f'(x)h + \frac{f''(x)}{2}h^2 + \frac{f'''(x)}{3!}h^3 + O(h^4)\\
    f(x - h) = f(x) - f'(x)h + \frac{f''(x)}{2}h^2 - \frac{f'''(x)}{3!}h^3 + O(h^4)
 \end{cases}$$
 Сложим уравнения:
 $$f(x+h) + f(x-h) = 2f(x) + f''(x)h^2 + O(h^4)$$
 $$f''(x) = \frac{f(x+h) - 2f(x) + f(x+h)}{h^2} + O(h^4)$$

 $O(h^4) :$

  $$
 \begin{cases}
    f(x + 2h) = f(x) + 2f'(x)h + \frac{f''(x)}{2}4h^2 + \frac{f'''(x)}{3!}8h^3 + \frac{f^{IV}(x)}{4!}16h^4 + \frac{f^{V}(x)}{5!}32h^5 + O(h^6)\\
    f(x + h) = f(x) + f'(x)h + \frac{f''(x)}{2}h^2 + \frac{f'''(x)}{3!}h^3 + \frac{f^{IV}(x)}{4!}h^4 + \frac{f^{V}(x)}{5!}h^5 + O(h^6)\\
    f(x - h) = f(x) - f'(x)h + \frac{f''(x)}{2}h^2 - \frac{f'''(x)}{3!}h^3 + \frac{f^{IV}(x)}{4!}h^4 - \frac{f^{V}(x)}{5!}h^5 + O(h^6)\\
    f(x - 2h) = f(x) - 2f'(x)h + \frac{f''(x)}{2}4h^2 - \frac{f'''(x)}{3!}8h^3 +\frac{f^{IV}(x)}{4!}16h^4 - \frac{f^{V}(x)}{5!}32h^5 + O(h^6)\\
 \end{cases}$$

 Нужно чтобы сократились первые, третьи, четвёртые и пятые производные

 Сложим попарно разложения по $h$ и по $2h$:

 $$
 \begin{cases}
    f(x+2h) + f(x-2h) = 2f(x) + 4h^2f''(x) + 32 \frac{f^{IV}(x)}{4!}h^4 + O(h^6)\\
    f(x+h) + f(x-h) = 2f(x) + h^2f''(x) + 2\frac{f^{IV}(x)}{4!}h^4 + O(h^6)
 \end{cases}$$
$$
 f(x+2h) - 16f(x+h) - 16f(x-h) + f(x-2h) = -30f(x) -12f''(x)h^2 +O(h^6)
 $$
