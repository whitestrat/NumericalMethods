# Численное интегрирование

Требуется решить уравнение:
$$u'' + \frac{1}{1+x}u' - \frac{x}{1+x}u = -\frac{x\ln{(1+x)}}{1+x}$$
С начальными условиями:
$$u(0) = 2; \quad u'(0)= 1$$
И сравнить полученный результат с аналитическим решением:
$$u_{а}(x) = \ln(1+x) + 2e^{-x}$$

0. Нормальная форма <br>
Замена $u = y^0; \quad u' = y^1; \quad u'' = y^2$
\begin{cases}
(y^{0})' = y^1 \\
(y^1)' = y^2 \\
y^2 = \frac{x}{1+x}y^0  - \frac{1}{1+x}y^1 - \frac{x\ln{(1+x)}}{1+x} = f(x, y_0, y_1)
\end{cases}

1. Метод Эйлера <br>
$$
\begin{cases}
y^{2}_{i} = f(x_{i}, y^{0}_{i}, y^{1}_{i}) \\
y^{1}_{i+1} =  y^{2}_{i}\Delta x \\
y^{0}_{i+1} = y^{1}_{i}\Delta x \\
\end{cases}
$$
Берём начальные данные и далее движемся по формулам выше
